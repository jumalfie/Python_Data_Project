# Calculate Projected salary next year

* Senior roles assume 5%
* Other roles assume 3%

```python
df_salary['df_salary_inflated'] = df_salary['salary_year_avg'] * 1.03

df_salary[['salary_year_avg', 'salary_year_inflated']]
```

```python
def projected_salary(row):
    if 'Senior' in row['job_title_short']:
        return 1.05 * row['salary_year_avg']
    else:
        return 1.03 * row['salary_year_avg']

df['salary_year_inflated'] = df_salary.apply(projected_salary, axis=1)

df_salary[['job_title_short', 'salary_year_avg', 'salary_year_inflated']]
```

```python
df['job_posted_date_str'] = df['job_posted_date'].apply(lambda x: x.strftime('%Y-%m-%d'))

df[['job_posted_date', 'job_posted_date_str']].head()
```

```python
from datetime import datetime 

current_date = datetime.now()
df['days_since_posted'] = df['job_posted_date'].apply(lambda date: (current_date - date).days)
df[['job_posted_date', 'days_since_posted']].head()

```

```python
df_filtered = df.dropna(subset=['salary_year_avg']).copy()

def salary_category(salary):
    if salary < 60_000:
        return 'Low'
    elif 60_000 <= salary <= 100_000:
        return 'Medium'
    else:
       return 'High'

df_filtered['salary_category'] = df['salary_year_avg'].apply(salary_category)

# df_filtered['salary_category'] = df['salary_year_avg'].apply(lambda x: 'Low' if x < 60_000 else 'Medium' if 60_000 <= x >=100_000 else 'High')
df_filtered


```
