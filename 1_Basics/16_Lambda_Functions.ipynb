```python
salaries = [95000, 120000, 105000, 90000, 130000]

average_salary = lambda salaries: sum(salaries)/len(salaries)
average_salary(salaries)
```

```python
job_titles = ['Data Scientist', 'Data Engineer', 'Machine Learning Engineer', 'Data Analyst']
list(filter(lambda data: 'Data' in data, job_titles))
```

```python
job_postings = [
    {'title': 'Data Scientist', 'skills': ['Python', 'SQL'], 'remote': True},
    {'title': 'Data Analyst', 'skills': ['Excel', 'SQL'], 'remote': False},
    {'title': 'Machine Learning Engineer', 'skills': ['Python', 'TensorFlow'], 'remote': True},
    {'title': 'Software Developer', 'skills': ['Java', 'C++'], 'remote': True}
]

list(filter(lambda job: 'Python' in job['skills'] and job['remote'], job_postings))
```

```python
employees = [
    {'name': 'Alice', 'department': 'IT', 'remote': True},
    {'name': 'Bob', 'department': 'Finance', 'remote': False},
    {'name': 'Charlie', 'department': 'HR', 'remote': False},
    {'name': 'David', 'department': 'IT', 'remote': True}
]

onsite_finance = list(filter(lambda non_remote: non_remote['remote'] == False and 'Finance' in non_remote['department'], employees))
print(onsite_finance)
```

```python
job_postings = [
    {'title': 'Data Scientist', 'skills': ['Python', 'SQL'], 'remote': True},
    {'title': 'Business Analyst', 'skills': ['Excel', 'SQL'], 'remote': True},
    {'title': 'Web Developer', 'skills': ['JavaScript', 'HTML'], 'remote': False},
    {'title': 'Database Administrator', 'skills': ['SQL', 'Oracle'], 'remote': True}
]

list(filter(lambda sql_jobs: 'SQL' in sql_jobs['skills'], job_postings))

```

```python
students = [
    {'name': 'John', 'grade': 75},
    {'name': 'Mary', 'grade': 88},
    {'name': 'James', 'grade': 92},
    {'name': 'Linda', 'grade': 67}
]

list(filter(lambda score: score['grade'] >= 80, students))
```

```python
for student in students:
  if student['grade'] >= 80:
    print(f'{student['name']} scored {student['grade']}.')
```

```python
books = [
    {'title': 'Python Basics', 'year': 2018},
    {'title': 'Data Science 101', 'year': 2020},
    {'title': 'Algorithms', 'year': 2013},
    {'title': 'AI Essentials', 'year': 2022}
]

list(filter(lambda publication_2015: 'Data Science 101' in publication_2015['title'] and publication_2015['year'] >= 2015, books))
```

```python
for book in books:
  if book['year'] >= 2015:
    print(f"{book['title']} was printed in {book['year']}.")
```
