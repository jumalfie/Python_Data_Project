```python
!pip install -q datasets

import pandas as pd
from datasets import load_dataset

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Clean up
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df['job_country'].value_counts().head(20)
```

```python
us_jobs = df[df['job_country'] == 'United States']
```

```python
us_jobs = us_jobs[us_jobs['salary_year_avg'].notna()]

```

```python
us_jobs.groupby('job_title_short')['salary_year_avg'].agg(['min', 'median', 'max', 'count']).sort_values(by='median', ascending=False)
```
