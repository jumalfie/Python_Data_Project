```python
job_counts = df.job_title_short.value_counts()
```

```python
country_counts = df.job_country.value_counts().head(5)
country_counts
```

```python
plt.bar(country_counts.index, country_counts, color='orange')
plt.xlabel('Country')
plt.ylabel('Number of Postings')
plt.title('Number of Jobs per Country')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
job_salary = df.groupby('job_title_short')['salary_year_avg'].mean()

job_salary
```

```python
job_salary = job_salary.sort_values(ascending=True)
plt.barh(job_salary.index, job_salary)
plt.xlabel('Average Salary')
plt.title('Average Salary by Job Title')
plt.show()
```

```python
monthly_salary = df.groupby('job_posted_month')['salary_year_avg'].mean()

monthly_salary
```

```python
plt.plot(monthly_salary.index, monthly_salary, color='red',label = 'Average Salary')
plt.xlabel('Month')
plt.ylabel('Average Salary')
plt.title('Average Salary by  job Month')
plt.legend()
plt.show()
```
