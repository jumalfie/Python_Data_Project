# Unpacking

```python
job_skills = ['python', 'excel', 'sql']
```

```python
skill1, skill2, skill3 = job_skills
```

```python
print(skill1)
print(skill2)
print(skill2)
```

```python
skill_concerned, *skills_not_concerned = job_skills
```

```python
print(skill_concerned)
print(skills_not_concerned)
```
